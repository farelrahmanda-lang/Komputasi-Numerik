```python
# ===== Eliminasi Gauss - Sistem Persamaan Linear 3 Variabel =====
# Tanda "# %%" memisahkan cell (dikenali Colab/Jupyter saat diimpor sebagai script)

# %% Cell 1: mengimpor library
import numpy as np  # numpy dipakai untuk operasi matriks dan array

# %% Cell 2: mendefinisikan sistem persamaan
# Matriks koefisien A (baris = persamaan, kolom = variabel x, y, z)
A = np.array([[ 2.0,  1.0, -1.0],   # persamaan 1: 2x + y - z
              [-3.0, -1.0,  2.0],   # persamaan 2: -3x - y + 2z
              [-2.0,  1.0,  2.0]])  # persamaan 3: -2x + y + 2z
# Vektor konstanta b (ruas kanan persamaan)
b = np.array([8.0, -11.0, -3.0])

# %% Cell 3: fungsi eliminasi Gauss
def gauss_elimination(A, b):
    n = len(b)                                  # jumlah variabel/persamaan
    M = np.hstack([A.copy(), b.reshape(-1, 1)]) # gabungkan A dan b menjadi matriks augmented [A|b]
    print("Matriks augmented awal:\n", M, "\n")  # tampilkan kondisi awal

    # ---- TAHAP 1: Eliminasi maju (membuat segitiga atas) ----
    for k in range(n - 1):                      # k = indeks kolom pivot
        p = np.argmax(np.abs(M[k:, k])) + k     # cari baris dengan nilai mutlak terbesar di kolom k
        if M[p, k] == 0:                        # jika pivot nol, sistem tidak punya solusi tunggal
            raise ValueError("Sistem tidak memiliki solusi tunggal")
        if p != k:                              # jika pivot terbesar bukan di baris k
            M[[k, p]] = M[[p, k]]               # tukar baris k dengan baris p
            print(f"Tukar baris {k+1} dengan baris {p+1}\n", M, "\n")
        for i in range(k + 1, n):               # untuk setiap baris di bawah pivot
            faktor = M[i, k] / M[k, k]          # hitung faktor pengali eliminasi
            M[i, k:] = M[i, k:] - faktor * M[k, k:]  # kurangi baris i agar elemen kolom k menjadi 0
            print(f"B{i+1} = B{i+1} - ({faktor:.4f}) * B{k+1}\n", M, "\n")

    # ---- TAHAP 2: Substitusi mundur ----
    x = np.zeros(n)                             # siapkan array penampung solusi
    for i in range(n - 1, -1, -1):              # mulai dari baris terakhir ke baris pertama
        jumlah = np.dot(M[i, i+1:n], x[i+1:n])  # jumlah suku variabel yang sudah diketahui
        x[i] = (M[i, -1] - jumlah) / M[i, i]    # hitung variabel ke-i
    return x                                    # kembalikan vektor solusi

# %% Cell 4: menjalankan fungsi dan menampilkan hasil
hasil = gauss_elimination(A, b)                 # panggil fungsi eliminasi Gauss
print("Solusi:")                                # judul hasil
print(f"x = {hasil[0]:.4f}")                    # nilai x
print(f"y = {hasil[1]:.4f}")                    # nilai y
print(f"z = {hasil[2]:.4f}")                    # nilai z

# %% Cell 5: verifikasi hasil
print("A @ x =", A @ hasil)                     # hasil kali A dan x harus sama dengan b
print("b     =", b)                             # vektor b sebagai pembanding
print("Cocok dengan numpy:", np.allclose(hasil, np.linalg.solve(A, b)))  # bandingkan dengan numpy
```
